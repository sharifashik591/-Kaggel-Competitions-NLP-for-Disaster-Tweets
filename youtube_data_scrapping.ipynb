```python
from playwright.async_api import async_playwright

VIDEO_ID = "HtNr_rp1juA"
CHAT_URL = f"https://www.youtube.com/live_chat?v={VIDEO_ID}"

async def scrape_live_chat():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox", "--disable-setuid-sandbox"]
        )

        context = await browser.new_context(
            user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                       "(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
        )

        page = await context.new_page()
        await page.goto(CHAT_URL, timeout=60000)
        await page.wait_for_timeout(15000)

        print("Live chat loaded. Listening...\n")

        seen = set()

        while True:
            messages = await page.query_selector_all("yt-live-chat-text-message-renderer")

            for msg in messages[-20:]:
                try:
                    author = await msg.query_selector("#author-name")
                    message = await msg.query_selector("#message")

                    author_text = await author.inner_text()
                    message_text = await message.inner_text()

                    key = author_text + message_text
                    if key not in seen:
                        seen.add(key)
                        print(f"{author_text}: {message_text}")
                except:
                    pass

            await page.wait_for_timeout(3000)


await scrape_live_chat()
```

```python
from playwright.async_api import async_playwright
import pandas as pd
from datetime import datetime

VIDEO_ID = "0mKXZEh43rQ"
CHAT_URL = f"https://www.youtube.com/live_chat?v={VIDEO_ID}"

df = pd.DataFrame(columns=["timestamp", "author", "message"])

async def scrape_live_chat():
    global df

    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox", "--disable-dev-shm-usage"]
        )

        context = await browser.new_context(
            viewport={"width":1280,"height":720},
            user_agent="Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                       "(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
        )

        page = await context.new_page()
        await page.goto(CHAT_URL, wait_until="domcontentloaded")
        await page.wait_for_timeout(15000)

        print("Live chat connected...\n")

        seen = set()

        while True:
          status_text = await page.inner_text("body")

          if "This live chat has ended" in status_text:
              print("🔴 Live stream ended.")
              break

          messages = await page.query_selector_all("yt-live-chat-text-message-renderer")

          for msg in messages[-20:]:
              try:
                  author = await msg.query_selector("#author-name")
                  message = await msg.query_selector("#message")

                  author_text = (await author.inner_text()).strip()
                  message_text = (await message.inner_text()).strip()

                  key = author_text + message_text
                  if key not in seen:
                      seen.add(key)

                      row = {
                          "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                          "author": author_text,
                          "message": message_text
                      }

                      df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)

                      print(f"{author_text}: {message_text}")

              except:
                  pass

          await page.wait_for_timeout(3000)


await scrape_live_chat()
```

```python
import pandas as pd
from transformers import pipeline


# Initialize multilingual sentiment model (English + Bangla)
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-xlm-roberta-base-sentiment"
)

# Function to analyze sentiment
def get_sentiment(text):
    if not text.strip():  # handle empty comments
        return "neutral", 0.0
    result = sentiment_pipeline(text[:512])[0]  # truncate to 512 chars
    return result["label"], round(result["score"], 3)

# Apply function to DataFrame
df[["sentiment", "confidence"]] = df["message"].apply(lambda x: pd.Series(get_sentiment(x)))

print(df)
```

```python
from transformers import pipeline
import pandas as pd


# Multilingual sentiment model
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-xlm-roberta-base-sentiment"
)

heading = "সরাসরি: শপথ নিলেন নবনির্বাচিত এমপিরা"

def context_sentiment(heading, comment):
    if not comment.strip():
        return "neutral", 0.0
    text = f"Heading: {heading}. Comment: {comment}"
    result = sentiment_pipeline(text[:512])[0]
    return result["label"], round(result["score"], 3)

df[["sentiment", "confidence"]] = df["message"].apply(
    lambda x: pd.Series(context_sentiment(heading, x))
)

print(df)
```
